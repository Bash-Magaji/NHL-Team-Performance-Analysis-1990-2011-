WEBSCRAPING HOCKEY TEAM DATASET

```python
from bs4 import BeautifulSoup
import requests
import pandas as pd
import time

url = "https://www.scrapethissite.com/pages/forms/"

all_rows = []
headers = None
page_num = 1

while True:
    params = {
        "page_num": page_num,
        "per_page": 100
    }

    page = requests.get(url, params=params, timeout=20)
    page.raise_for_status()

    soup = BeautifulSoup(page.text, "html.parser")

    # Get the table headings once
    if headers is None:
        table_head = soup.find_all("th")
        headers = [title.get_text(strip=True) for title in table_head]

    # Find the actual team rows
    rows = soup.select("tr.team")

    # Stop when no more records exist
    if not rows:
        break

    print(f"Scraping page {page_num}...")

    for row in rows:
        row_data = row.find_all("td")
        individual_row_data = [
            data.get_text(strip=True)
            for data in row_data
        ]

        all_rows.append(individual_row_data)

    page_num += 1
    time.sleep(1)

# Create DataFrame after scraping all pages
df = pd.DataFrame(all_rows, columns=headers)
```
